```python

# 4B. CBOW Model using Twitter Dataset

# 1. Import libraries
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt

from gensim.models import Word2Vec
from gensim.utils import simple_preprocess


# 2. Load dataset
data = pd.read_csv(r'D:\Downloads\LP IV Updated Datasets\LP IV Updated Datasets\CBOW\twitter_dataset.csv')

print("Dataset Shape:", data.shape)
print(data.head())

# Select tweet text
tweets = data['Text'].dropna().astype(str)


# 3. Clean and preprocess tweets
def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove mentions
    text = re.sub(r'@\w+', '', text)

    # Remove hashtag symbol but keep the word
    text = re.sub(r'#', '', text)

    # Tokenize text and remove punctuation
    tokens = simple_preprocess(text, deacc=True)

    return tokens


# Apply preprocessing
tokenized_tweets = [preprocess(tweet) for tweet in tweets]

# Remove empty tweets
tokenized_tweets = [t for t in tokenized_tweets if len(t) > 0]

print("\nSample Tokenized Tweets:")
print(tokenized_tweets[:3])


# 4. Train CBOW Word2Vec model
model = Word2Vec(
    sentences=tokenized_tweets,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    sg=0,
    epochs=20
)

print("\nCBOW Model Training Completed!")


# 5. Display vocabulary information
print("\nVocabulary Size:", len(model.wv))

print("\nFirst 10 Words:")
print(model.wv.index_to_key[:10])


# 6. Get word embedding
word = model.wv.index_to_key[0]

embedding = model.wv[word]

print("\nSelected Word:", word)
print("Word Embedding:")
print(embedding)

print("Embedding Shape:", embedding.shape)


# 7. Find similar words
similar_words = model.wv.most_similar(word, topn=5)

print("\nWords Similar to", word)

for w, score in similar_words:
    print(w, ":", score)


# 8. Calculate similarity between two words
word1 = model.wv.index_to_key[0]
word2 = model.wv.index_to_key[1]

similarity = model.wv.similarity(word1, word2)

print("\nSimilarity between", word1, "and", word2, ":", similarity)


# # 9. Save trained model
model.save("cbow_twitter.model")

print("\nModel Saved Successfully!")
```
