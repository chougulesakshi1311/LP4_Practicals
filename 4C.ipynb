```python
import re
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
```

```python
with open(r"D:\Downloads\LP IV Updated Datasets\LP IV Updated Datasets\CBOW\document3.txt", "r", encoding="utf-8") as file:
    text = file.read()
```

```python
print("Original Text: ", text[:500])
```

```python
text = text.lower()

sentences = re.split(r'[.!?]+', text)

tokenized_sentences = [
    simple_preprocess(sentence, deacc=True)
    for sentence in sentences
]

tokenized_sentences = [
    sentence for sentence in tokenized_sentences
    if len(sentences)>0
]

print("Tokenized sentences: ", tokenized_sentences[:5])
```

```python
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=50,
    workers=1, 
    epochs=100,
    sg=0,
    seed=42,
    min_count=1,
    window=3
)

print("CBOW model trained successfully!")
```

```python
print("Vocabulary size: ", len(model.wv))
```

```python
print("Vocabulary words: ")
print(model.wv.index_to_key[:10])
```

```python
word = "influenza"

if word in model.wv:
    embedding = model.wv[word]

    print("Selected word: ", word)
    print("Embedding: ", embedding)
    print("Embedding shape: ", embedding.shape)
```

```python
if word in model.wv:
    print("Words similar to: ", word)

    similar_words = model.wv.most_similar(
        word, topn=5
    )

    for w,score in similar_words:
        print(w, ":", score)
```

```python
word1 = "influenza"
word2 = "covid"

if word1 in model.wv and word2 in model.wv:
    similarity = model.wv.similarity(word1, word2)

    print("\nSimilarity between", word1, "and", word2)
    print("Cosine Similarity:", similarity)
```

```python
model.save("cbow_document.model")
print("Model saved successfully")
```
